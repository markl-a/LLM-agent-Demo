# MetaGPT 基礎概念

## 📚 簡介

MetaGPT 是一個創新的多 Agent 框架，它將軟體公司的組織結構和工作流程編碼為 AI Agents，實現從需求到代碼的全自動化開發。

### 核心理念：軟體公司 = LLM + SOP

MetaGPT 模擬真實軟體公司的運作方式：
- **產品經理 (ProductManager)**: 撰寫 PRD（產品需求文檔）
- **架構師 (Architect)**: 設計系統架構和技術方案
- **工程師 (Engineer)**: 編寫高質量代碼
- **測試工程師 (QA Engineer)**: 編寫測試用例和執行測試

所有 Agent 遵循標準作業流程（SOP），協同工作完成軟體開發。

## 3. 核心概念詳解

### 3.1 MetaGPT 的工作流程

```
用戶需求 
   ↓
產品經理 → PRD（產品需求文檔）
   ↓
架構師 → 系統設計文檔 + API設計 + 數據結構
   ↓
工程師 → 代碼實現
   ↓
測試工程師 → 測試用例 + 測試報告
   ↓
完整的軟體項目
```

### 3.2 關鍵組件

1. **Role（角色）**: 代表不同職位的 Agent，每個角色有特定的職責和行為
2. **Action（行為）**: Agent 執行的具體動作，如撰寫 PRD、編寫代碼
3. **Memory（記憶）**: Agent 的上下文記憶，存儲歷史信息
4. **Environment（環境）**: Agent 工作的環境，管理消息傳遞
5. **Message（消息）**: 角色間傳遞的結構化信息
6. **Schema（模式）**: 定義文檔和數據的結構

````python
# 檢查 MetaGPT 是否正確安裝
try:
    import metagpt
    print(f"✓ MetaGPT 版本: {metagpt.__version__}")
    print("✓ MetaGPT 已成功導入")
    
    # 導入核心模組
    from metagpt.roles import ProductManager, Architect, Engineer
    from metagpt.team import Team
    print("✓ 核心模組導入成功")
    
except ImportError as e:
    print(f"❌ 導入失敗: {e}")
    print("請運行: pip install metagpt")
````

````python
# 示例 1: 使用 Team 創建軟體開發團隊
import asyncio
from metagpt.team import Team

async def create_simple_project():
    """
    創建一個簡單的命令行工具項目
    
    注意: 這會調用多次 LLM API，需要一定時間和成本
    """
    # 定義項目需求
    idea = """
    創建一個簡單的命令行待辦事項管理工具（TODO CLI）:
    1. 支持添加待辦事項
    2. 支持列出所有待辦事項
    3. 支持標記待辦事項為完成
    4. 支持刪除待辦事項
    5. 使用 JSON 文件存儲數據
    """
    
    # 創建團隊（默認包含 PM、Architect、Engineer）
    team = Team()
    
    # 設置投資金額（控制生成質量和成本）
    team.invest(investment=3.0)  # 3 美元預算
    
    # 運行項目開發
    team.run_project(idea)
    
    # 等待完成
    await team.run(n_round=5)  # 最多運行 5 輪
    
    print("\n✓ 項目開發完成！")
    print(f"生成的文件保存在: {team.env.workspace.path}")

# 運行示例（取消註釋下面的代碼來執行）
# 注意：這會實際調用 LLM API 並產生費用
# await create_simple_project()

print("提示: 取消註釋上面的代碼來運行完整示例")
print("⚠️  警告: 運行會調用 LLM API 並產生費用（約 1-3 美元）")
````

## 7. 消息傳遞機制

理解 MetaGPT 中角色之間如何通過消息進行通信。

````python
from metagpt.schema import Message
from metagpt.actions import Action

# 創建一個消息示例
message = Message(
    content="創建一個計算器應用",
    role="User",
    cause_by=Action  # 由某個 Action 觸發
)

print("消息結構:")
print(f"  內容: {message.content}")
print(f"  發送者: {message.role}")
print(f"  類型: {type(message).__name__}")

# 消息傳遞流程示例
print("\n消息傳遞流程:")
workflow = [
    ("User", "ProductManager", "需求描述"),
    ("ProductManager", "Architect", "PRD 文檔"),
    ("Architect", "Engineer", "設計文檔"),
    ("Engineer", "QA", "代碼實現"),
    ("QA", "User", "測試報告")
]

for from_role, to_role, content in workflow:
    print(f"  {from_role:20s} → {to_role:20s} : {content}")
````

## 8. 簡化示例：單個角色運行

不需要運行完整流程，也可以單獨使用某個角色。

````python
# 示例: 只使用產品經理生成 PRD
async def generate_prd_only():
    """
    只使用產品經理角色生成 PRD 文檔
    """
    from metagpt.roles import ProductManager
    from metagpt.logs import logger
    
    # 創建產品經理
    pm = ProductManager()
    
    # 定義需求
    requirement = """
    設計一個在線書籍管理系統:
    - 用戶可以瀏覽書籍
    - 用戶可以搜索書籍
    - 用戶可以添加書籍到收藏
    - 管理員可以添加、編輯、刪除書籍
    """
    
    # 運行產品經理角色
    result = await pm.run(requirement)
    
    print("\nPRD 文檔已生成:")
    print(result)
    
    return result

# 取消註釋以運行
# prd = await generate_prd_only()

print("提示: 這個示例展示了如何單獨使用一個角色")
print("取消註釋以執行（會調用 LLM API）")
````

## 10. 配置管理

## 13. MetaGPT 的優勢和限制

### ✅ 優勢

1. **完整的開發流程**
   - 從需求到代碼的全流程自動化
   - 生成完整的文檔和代碼
   - 遵循軟體工程最佳實踐

2. **多角色協作**
   - 模擬真實軟體團隊的協作方式
   - 每個角色專注於自己的職責
   - 通過結構化消息進行溝通

3. **結構化輸出**
   - 生成規範的 PRD、設計文檔
   - 代碼遵循最佳實踐
   - 易於理解和維護

4. **可擴展性**
   - 支持自定義角色和行為
   - 可以集成外部工具
   - 靈活的配置選項

### ⚠️ 限制

1. **成本較高**
   - 需要多次 LLM API 調用
   - 完整流程成本 1-30 美元不等
   - 大型項目成本更高

2. **時間消耗**
   - 完整流程需要幾分鐘到幾十分鐘
   - 不適合實時應用
   - 需要耐心等待

3. **學習曲線**
   - 概念相對複雜
   - 配置需要一定經驗
   - 需要理解軟體工程流程

4. **適用範圍**
   - 主要適合中小型項目
   - 不適合高度定制化的項目
   - 生成的代碼仍需人工審核
