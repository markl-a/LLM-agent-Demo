## 案例 1: 內容創作工作流

### 場景描述
為技術博客創建一篇關於 AI 技術的文章，從主題選擇、研究、撰寫到編輯的完整流程。

### 團隊組成
- **內容策略師**: 選擇熱門主題和關鍵字
- **研究員**: 收集最新資訊和數據
- **作家**: 撰寫文章內容
- **編輯**: 審校和優化文章

```python
# 案例 1: 內容創作工作流

# 1. 內容策略師
content_strategist = Agent(
    role='內容策略師',
    goal='識別熱門主題和優化 SEO 策略',
    backstory="""你是一位經驗豐富的內容策略專家，
    擅長分析市場趨勢、識別熱門話題，並制定內容策略。
    你了解 SEO 最佳實踐，能夠為文章選擇合適的關鍵字。""",
    verbose=True,
    allow_delegation=False
)

# 2. 研究員
content_researcher = Agent(
    role='技術研究員',
    goal='收集準確的技術資訊和最新趨勢',
    backstory="""你是一位技術研究專家，能夠快速找到可靠的資訊來源，
    理解複雜的技術概念，並整理成易於理解的內容。""",
    verbose=True,
    allow_delegation=False
)

# 3. 作家
content_writer = Agent(
    role='技術作家',
    goal='撰寫引人入勝且易於理解的技術文章',
    backstory="""你是一位專業的技術作家，擅長將複雜的技術概念
    轉化為通俗易懂的語言。你的文章結構清晰、邏輯嚴密，
    同時又不失趣味性。""",
    verbose=True,
    allow_delegation=False
)

# 4. 編輯
content_editor = Agent(
    role='內容編輯',
    goal='確保文章質量、準確性和可讀性',
    backstory="""你是一位細心的編輯，擅長發現文章中的錯誤、
    改善文章結構、優化表達方式。你確保每篇文章都達到
    出版標準。""",
    verbose=True,
    allow_delegation=False
)

print("內容創作團隊已組建！")
```

```python
# 定義任務

# 任務 1: 主題策劃
strategy_task = Task(
    description="""分析 2024 年 AI 領域的熱門話題，選擇一個適合
    技術博客的主題。請考慮：
    1. 話題的熱度和搜索量
    2. 目標受眾的興趣
    3. SEO 優化關鍵字
    4. 內容的獨特性
    
    提供：主題名稱、目標關鍵字列表、文章角度建議""",
    agent=content_strategist,
    expected_output="包含主題、關鍵字和內容角度的策略報告"
)

# 任務 2: 研究
research_task = Task(
    description="""基於選定的主題，進行深入研究。請收集：
    1. 技術背景和基本概念
    2. 最新發展和趨勢
    3. 實際應用案例
    4. 專家觀點和數據統計
    5. 相關的圖表和視覺化建議
    
    整理成結構化的研究報告。""",
    agent=content_researcher,
    expected_output="詳細的研究報告，包含所有必要資訊",
    context=[strategy_task]
)

# 任務 3: 撰寫
writing_task = Task(
    description="""基於研究結果，撰寫一篇完整的博客文章。要求：
    1. 引人入勝的標題和開頭
    2. 清晰的文章結構（引言、主體、結論）
    3. 使用簡潔易懂的語言
    4. 包含具體例子和應用場景
    5. 自然融入 SEO 關鍵字
    6. 字數約 1500-2000 字
    
    文章應該既專業又易讀。""",
    agent=content_writer,
    expected_output="完整的博客文章初稿",
    context=[strategy_task, research_task]
)

# 任務 4: 編輯
editing_task = Task(
    description="""審校和優化文章。請檢查：
    1. 語法、拼寫和標點符號
    2. 文章結構和邏輯流程
    3. 技術準確性
    4. 可讀性和吸引力
    5. SEO 優化效果
    
    提供最終版本和改進建議摘要。""",
    agent=content_editor,
    expected_output="最終定稿文章和編輯報告",
    context=[writing_task]
)

print("任務已定義！")
```

```python
# 組建內容創作團隊
content_crew = Crew(
    agents=[content_strategist, content_researcher, content_writer, content_editor],
    tasks=[strategy_task, research_task, writing_task, editing_task],
    process=Process.sequential,
    verbose=2
)

# 執行工作流
print("開始內容創作流程...\n")
print("="*80)

result = content_crew.kickoff()

print("\n" + "="*80)
print("內容創作完成！\n")
print("最終文章：")
print(result)
```

## 案例 2: 市場研究團隊

### 場景描述
為一家 AI 創業公司進行市場調研，分析競爭對手和市場機會。

### 團隊組成
- **市場分析師**: 分析整體市場趨勢
- **競品研究員**: 研究競爭對手
- **用戶研究員**: 分析用戶需求
- **戰略顧問**: 提供戰略建議

```python
# 案例 2: 市場研究團隊

# 市場分析師
market_analyst = Agent(
    role='市場分析師',
    goal='分析 AI SaaS 市場的整體趨勢和規模',
    backstory="""你是一位資深市場分析師，擅長使用數據分析工具，
    能夠從宏觀角度理解市場動態，識別增長機會和潛在風險。""",
    verbose=True
)

# 競品研究員
competitor_researcher = Agent(
    role='競品研究員',
    goal='深入分析主要競爭對手的產品和策略',
    backstory="""你是競品分析專家，善於收集競爭對手的公開信息，
    分析他們的產品特點、定價策略、市場定位和優劣勢。""",
    verbose=True
)

# 用戶研究員
user_researcher = Agent(
    role='用戶研究員',
    goal='理解目標用戶的需求、痛點和期望',
    backstory="""你是用戶體驗研究專家，擅長通過各種方法了解用戶需求，
    包括用戶訪談、問卷調查、行為分析等。""",
    verbose=True
)

# 戰略顧問
strategy_consultant = Agent(
    role='戰略顧問',
    goal='基於市場研究提供可行的戰略建議',
    backstory="""你是商業戰略顧問，擅長整合各類市場信息，
    提供切實可行的商業策略和行動計劃。""",
    verbose=True
)

print("市場研究團隊已組建！")
```

```python
# 定義市場研究任務

market_analysis_task = Task(
    description="""分析 AI 輔助寫作工具的市場現狀。請研究：
    1. 市場規模和增長率
    2. 主要市場趨勢
    3. 目標客戶群體
    4. 市場細分情況
    5. 未來發展預測""",
    agent=market_analyst,
    expected_output="市場分析報告"
)

competitor_analysis_task = Task(
    description="""分析 AI 寫作工具領域的主要競爭對手（如 Jasper, Copy.ai, Writesonic）。
    請包括：
    1. 產品功能對比
    2. 定價策略
    3. 目標用戶群
    4. 優勢和劣勢
    5. 市場份額估計""",
    agent=competitor_researcher,
    expected_output="競品分析報告"
)

user_research_task = Task(
    description="""研究潛在用戶對 AI 寫作工具的需求。分析：
    1. 主要使用場景
    2. 核心需求和痛點
    3. 對現有產品的評價
    4. 願意付費的功能
    5. 使用障礙和顧慮""",
    agent=user_researcher,
    expected_output="用戶需求報告"
)

strategy_task = Task(
    description="""基於市場、競品和用戶研究，提供戰略建議。包括：
    1. 市場定位建議
    2. 差異化策略
    3. 產品功能優先級
    4. 定價策略建議
    5. Go-to-Market 計劃
    6. 風險和挑戰""",
    agent=strategy_consultant,
    expected_output="完整的戰略建議報告",
    context=[market_analysis_task, competitor_analysis_task, user_research_task]
)

print("市場研究任務已定義！")
```

```python
# 組建市場研究團隊
market_crew = Crew(
    agents=[market_analyst, competitor_researcher, user_researcher, strategy_consultant],
    tasks=[market_analysis_task, competitor_analysis_task, user_research_task, strategy_task],
    process=Process.sequential,
    verbose=2
)

# 執行市場研究
print("開始市場研究...\n")
result = market_crew.kickoff()
print("\n市場研究完成！\n")
print(result)
```

## 案例 3: 客戶服務團隊

### 場景描述
構建智能客服系統，自動處理客戶諮詢和問題。

### 團隊組成
- **分類專員**: 分析和分類客戶問題
- **知識庫專員**: 查找相關解決方案
- **客服專員**: 提供個性化回覆
- **質量專員**: 審核回覆質量

```python
# 案例 3: 客戶服務團隊

# 問題分類專員
classifier = Agent(
    role='問題分類專員',
    goal='準確分類客戶問題和確定優先級',
    backstory="""你是客服系統的第一線，擅長快速理解客戶問題的本質，
    將問題分類到正確的類別，並判斷問題的緊急程度。""",
    verbose=True
)

# 知識庫專員
knowledge_base_agent = Agent(
    role='知識庫專員',
    goal='從知識庫中找到最相關的解決方案',
    backstory="""你精通公司的產品和服務，熟悉所有文檔和常見問題解答，
    能夠快速找到準確的答案。""",
    verbose=True
)

# 客服專員
support_agent = Agent(
    role='客服專員',
    goal='提供友好、專業、有幫助的客戶回覆',
    backstory="""你是經驗豐富的客服人員，擅長用友好的語氣溝通，
    能夠將技術信息轉化為客戶易懂的語言，同時保持專業性。""",
    verbose=True
)

# 質量保證專員
qa_agent = Agent(
    role='質量保證專員',
    goal='確保客戶回覆的準確性和專業性',
    backstory="""你負責審核所有客戶回覆，確保信息準確、語氣恰當、
    完全解決了客戶的問題。""",
    verbose=True
)

print("客戶服務團隊已組建！")
```

```python
# 模擬客戶問題
customer_inquiry = """
主題：無法登入帳戶
客戶：張先生
問題：我已經嘗試重置密碼三次了，但仍然無法登入我的帳戶。
每次輸入新密碼後，系統都說密碼錯誤。這個問題已經持續兩天了，
嚴重影響了我的工作。請盡快幫我解決！
"""

# 定義客服任務
classify_task = Task(
    description=f"""分析以下客戶問題：
    {customer_inquiry}
    
    請提供：
    1. 問題類別（技術問題/帳號問題/計費問題/功能諮詢/其他）
    2. 優先級（低/中/高/緊急）
    3. 問題摘要
    4. 關鍵詞""",
    agent=classifier,
    expected_output="問題分類和優先級報告"
)

knowledge_search_task = Task(
    description="""基於分類結果，從知識庫搜索相關解決方案。
    
    需要找到：
    1. 類似問題的解決方案
    2. 相關的技術文檔
    3. 常見的解決步驟
    4. 可能的原因分析""",
    agent=knowledge_base_agent,
    expected_output="相關解決方案和資源列表",
    context=[classify_task]
)

response_task = Task(
    description=f"""為客戶撰寫回覆郵件。
    
    原始問題：{customer_inquiry}
    
    回覆要求：
    1. 友好且專業的語氣
    2. 對客戶的困擾表示理解
    3. 提供清晰的解決步驟
    4. 包含替代方案（如果有）
    5. 提供後續聯繫方式""",
    agent=support_agent,
    expected_output="完整的客戶回覆郵件",
    context=[classify_task, knowledge_search_task]
)

qa_task = Task(
    description="""審核客戶回覆。檢查：
    1. 是否完全回答了客戶的問題
    2. 解決方案是否準確可行
    3. 語氣是否恰當
    4. 是否有語法錯誤
    5. 是否需要改進
    
    如果有問題，提供改進建議。""",
    agent=qa_agent,
    expected_output="審核報告和最終版本回覆",
    context=[response_task]
)

print("客服任務已定義！")
```

```python
# 組建客服團隊
support_crew = Crew(
    agents=[classifier, knowledge_base_agent, support_agent, qa_agent],
    tasks=[classify_task, knowledge_search_task, response_task, qa_task],
    process=Process.sequential,
    verbose=2
)

# 執行客服流程
print("處理客戶問題...\n")
result = support_crew.kickoff()
print("\n客戶回覆已生成！\n")
print(result)
```

## 案例 4: 數據分析團隊

### 場景描述
分析公司的銷售數據，生成洞察報告。

### 團隊組成
- **數據工程師**: 清理和準備數據
- **數據分析師**: 執行統計分析
- **數據科學家**: 發現深層洞察
- **報告撰寫員**: 生成可視化報告

```python
# 案例 4: 數據分析團隊

# 數據工程師
data_engineer = Agent(
    role='數據工程師',
    goal='清理、轉換和準備數據以供分析',
    backstory="""你是數據工程專家，擅長數據清理、ETL 流程和數據質量控制。
    你確保數據分析師獲得乾淨、結構化的數據。""",
    verbose=True
)

# 數據分析師
data_analyst = Agent(
    role='數據分析師',
    goal='執行統計分析並識別關鍵指標',
    backstory="""你是數據分析專家，精通統計學和數據分析工具。
    你能夠從數據中提取有價值的見解，並用數據講故事。""",
    verbose=True
)

# 數據科學家
data_scientist = Agent(
    role='數據科學家',
    goal='發現深層模式和預測性洞察',
    backstory="""你是數據科學專家，擅長機器學習和高級分析。
    你能夠發現隱藏的模式，並提供預測性見解。""",
    verbose=True
)

# 報告作家
report_writer = Agent(
    role='業務報告撰寫員',
    goal='將分析結果轉化為易懂的業務報告',
    backstory="""你擅長將複雜的數據分析結果轉化為清晰的業務洞察，
    創建視覺化圖表，並撰寫高管可以理解的報告。""",
    verbose=True
)

print("數據分析團隊已組建！")
```

```python
# 模擬數據場景
data_context = """
數據源：2024 年 Q1-Q3 電商平台銷售數據
包含欄位：
- 訂單ID、日期、客戶ID、產品類別、銷售額、數量、地區、行銷渠道

業務問題：
1. 哪些產品類別表現最好？
2. 銷售趨勢如何變化？
3. 不同行銷渠道的 ROI 如何？
4. 有哪些可以改進的領域？
"""

# 定義數據分析任務
data_prep_task = Task(
    description=f"""{data_context}
    
    執行數據準備工作：
    1. 檢查數據完整性和質量
    2. 識別並處理缺失值和異常值
    3. 數據格式標準化
    4. 創建分析所需的衍生欄位
    5. 數據質量報告""",
    agent=data_engineer,
    expected_output="數據準備報告和清理後的數據摘要"
)

analysis_task = Task(
    description="""對準備好的數據執行描述性統計分析：
    1. 計算關鍵指標（總銷售額、平均訂單價值、客戶獲取成本等）
    2. 分析產品類別表現
    3. 時間序列趨勢分析
    4. 地區和渠道表現對比
    5. 識別異常和極端值""",
    agent=data_analyst,
    expected_output="統計分析報告和關鍵發現",
    context=[data_prep_task]
)

insights_task = Task(
    description="""深入分析數據，發現洞察：
    1. 客戶細分和行為模式
    2. 產品關聯性分析
    3. 銷售預測（Q4）
    4. 識別增長機會
    5. 風險因素分析""",
    agent=data_scientist,
    expected_output="深度洞察和預測報告",
    context=[data_prep_task, analysis_task]
)

report_task = Task(
    description="""創建高管摘要報告：
    1. 執行摘要（關鍵發現和建議）
    2. 業務表現總覽
    3. 詳細分析結果
    4. 可視化圖表描述
    5. 行動建議
    6. 附錄（詳細數據）
    
    報告應該清晰、簡潔、面向業務決策。""",
    agent=report_writer,
    expected_output="完整的業務分析報告",
    context=[data_prep_task, analysis_task, insights_task]
)

print("數據分析任務已定義！")
```

```python
# 組建數據分析團隊
analytics_crew = Crew(
    agents=[data_engineer, data_analyst, data_scientist, report_writer],
    tasks=[data_prep_task, analysis_task, insights_task, report_task],
    process=Process.sequential,
    verbose=2
)

# 執行數據分析
print("開始數據分析...\n")
result = analytics_crew.kickoff()
print("\n數據分析完成！\n")
print(result)
```

## 案例 5: 軟體開發團隊

### 場景描述
完整的軟體開發流程，從需求分析到測試。

### 團隊組成
- **產品經理**: 定義需求和用戶故事
- **架構師**: 設計系統架構
- **開發者**: 實現功能
- **測試工程師**: 編寫測試計劃

```python
# 案例 5: 軟體開發團隊

product_manager = Agent(
    role='產品經理',
    goal='定義清晰的產品需求和用戶故事',
    backstory="""你是經驗豐富的產品經理，擅長理解用戶需求，
    定義產品功能，並將業務目標轉化為技術需求。""",
    verbose=True
)

architect = Agent(
    role='系統架構師',
    goal='設計可擴展、可維護的系統架構',
    backstory="""你是資深系統架構師，精通各種架構模式和最佳實踐，
    能夠設計健壯、可擴展的系統。""",
    verbose=True
)

developer = Agent(
    role='全棧開發者',
    goal='實現高質量、可維護的代碼',
    backstory="""你是經驗豐富的全棧開發者，精通前後端技術，
    編寫乾淨、高效的代碼，遵循最佳實踐。""",
    verbose=True
)

qa_engineer = Agent(
    role='QA 測試工程師',
    goal='確保軟體質量和可靠性',
    backstory="""你是細心的測試工程師，擅長設計測試用例，
    發現潛在問題，確保軟體達到質量標準。""",
    verbose=True
)

print("軟體開發團隊已組建！")
```

```python
# 項目需求
project_brief = """
項目：待辦事項管理 Web 應用

目標用戶：個人和小團隊

核心需求：
- 用戶可以創建、編輯、刪除待辦事項
- 支持分類和標籤
- 設置優先級和截止日期
- 簡潔美觀的用戶界面
- 跨設備同步
"""

# 需求分析任務
requirements_task = Task(
    description=f"""{project_brief}
    
    請完成以下工作：
    1. 詳細的功能需求列表
    2. 用戶故事（User Stories）
    3. 驗收標準
    4. 功能優先級排序
    5. 非功能需求（性能、安全等）""",
    agent=product_manager,
    expected_output="產品需求文檔（PRD）"
)

# 架構設計任務
architecture_task = Task(
    description="""基於產品需求，設計系統架構：
    1. 技術棧選擇（前端、後端、數據庫）
    2. 系統架構圖
    3. 數據模型設計
    4. API 設計
    5. 安全性考慮
    6. 可擴展性設計""",
    agent=architect,
    expected_output="技術設計文檔",
    context=[requirements_task]
)

# 開發任務
development_task = Task(
    description="""根據架構設計，提供實現方案：
    1. 項目結構
    2. 核心功能實現概要（偽代碼或描述）
    3. 前端組件設計
    4. 後端 API 實現
    5. 數據庫 Schema
    6. 關鍵代碼片段示例""",
    agent=developer,
    expected_output="開發實現方案",
    context=[requirements_task, architecture_task]
)

# 測試任務
testing_task = Task(
    description="""制定完整的測試計劃：
    1. 測試策略
    2. 單元測試計劃
    3. 集成測試計劃
    4. E2E 測試場景
    5. 性能測試計劃
    6. 安全測試檢查清單
    7. 測試用例示例""",
    agent=qa_engineer,
    expected_output="測試計劃文檔",
    context=[requirements_task, architecture_task, development_task]
)

print("開發任務已定義！")
```

```python
# 組建開發團隊
dev_crew = Crew(
    agents=[product_manager, architect, developer, qa_engineer],
    tasks=[requirements_task, architecture_task, development_task, testing_task],
    process=Process.sequential,
    verbose=2
)

# 執行開發流程
print("開始軟體開發流程...\n")
result = dev_crew.kickoff()
print("\n開發計劃完成！\n")
print(result)
```

## 案例 6: 社交媒體管理團隊

### 場景描述
為品牌管理社交媒體，包括內容策劃、創作和發布。

### 團隊組成
- **社交媒體策略師**: 制定內容策略
- **創意總監**: 構思創意概念
- **文案撰寫**: 撰寫貼文內容
- **社群經理**: 優化發布策略

```python
# 案例 6: 社交媒體管理團隊

social_strategist = Agent(
    role='社交媒體策略師',
    goal='制定有效的社交媒體內容策略',
    backstory="""你是社交媒體營銷專家，了解各平台的特點和算法，
    能夠制定提高參與度和觸及率的內容策略。""",
    verbose=True
)

creative_director = Agent(
    role='創意總監',
    goal='構思獨特吸引人的創意概念',
    backstory="""你是創意專家，擅長產生原創想法，
    設計引人注目的視覺和文字內容概念。""",
    verbose=True
)

copywriter = Agent(
    role='社交媒體文案',
    goal='撰寫吸引人的社交媒體貼文',
    backstory="""你是社交媒體文案專家，擅長寫簡潔有力的文字，
    了解不同平台的語言風格，能夠激發用戶互動。""",
    verbose=True
)

community_manager = Agent(
    role='社群經理',
    goal='優化內容並規劃發布時程',
    backstory="""你是社群管理專家，了解最佳發布時間，
    hashtag 策略，以及如何最大化內容的影響力。""",
    verbose=True
)

print("社交媒體團隊已組建！")
```

```python
# 活動簡報
campaign_brief = """
品牌：環保科技新創公司
產品：AI 驅動的智能回收箱
目標：提高品牌知名度，獲得早期用戶
平台：Instagram, Twitter, LinkedIn
時間：下週一產品發布
"""

# 策略任務
strategy_social_task = Task(
    description=f"""{campaign_brief}
    
    制定社交媒體活動策略：
    1. 目標受眾分析
    2. 關鍵訊息和價值主張
    3. 內容主題方向
    4. KPI 指標設定
    5. 各平台策略差異""",
    agent=social_strategist,
    expected_output="社交媒體活動策略文檔"
)

# 創意任務
creative_task = Task(
    description="""基於策略，構思創意概念：
    1. 3-5 個內容創意方向
    2. 視覺風格建議
    3. 互動元素設計（問答、投票等）
    4. Hashtag 創意
    5. 用戶參與機制""",
    agent=creative_director,
    expected_output="創意概念提案",
    context=[strategy_social_task]
)

# 文案任務
copywriting_task = Task(
    description="""為產品發布撰寫社交媒體貼文：
    
    請為以下平台各撰寫 2 則貼文：
    1. Instagram（配圖文字、Caption、Hashtags）
    2. Twitter（280 字以內）
    3. LinkedIn（專業語氣）
    
    要求：
    - 引人注目的開頭
    - 清晰的價值主張
    - 行動呼籲（CTA）""",
    agent=copywriter,
    expected_output="完整的社交媒體貼文集",
    context=[strategy_social_task, creative_task]
)

# 發布計劃任務
publishing_task = Task(
    description="""制定內容發布計劃：
    1. 發布時程表（時間、平台、內容）
    2. Hashtag 策略
    3. 互動回應指南
    4. A/B 測試建議
    5. 監測指標和調整策略""",
    agent=community_manager,
    expected_output="社交媒體發布計劃",
    context=[strategy_social_task, creative_task, copywriting_task]
)

print("社交媒體任務已定義！")
```

```python
# 組建社交媒體團隊
social_crew = Crew(
    agents=[social_strategist, creative_director, copywriter, community_manager],
    tasks=[strategy_social_task, creative_task, copywriting_task, publishing_task],
    process=Process.sequential,
    verbose=2
)

# 執行社交媒體活動策劃
print("開始社交媒體活動策劃...\n")
result = social_crew.kickoff()
print("\n活動策劃完成！\n")
print(result)
```
